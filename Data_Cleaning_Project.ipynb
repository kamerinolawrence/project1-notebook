# Customer Churn Data Cleaning and Preparation

## Project Overview

Customer Churn occurs when customers stop using a company's products or services.
This project focuses on cleaning and preparing a Telecommunicatins customer dataset for further analysis.
The main goal is to identify and address data quality issues such as 
Missing Values

Duplicate Records

Incorrect Data Types

Inconsistent Values 

Potential Outliers 

The cleaned dataset will provide a reliable foundation for subsequent Customer churn anlaysis.


## Busines  Problem 

Telecommunication companies need reliable customer data to understand customer behaviour and identify factors associated with customer churn 

Before performing analysis , the dataset must be examined and cleaned to ensure that the results are accurate and trustworthy.

This project therefore focuses on systematically assessing and improving the quality of the customer churn dataset

## Project Objectives 
The objectives of thus projects are to :

1. Understand the structure and characteristics of the raw dataset
2. Assess the quality of the data
3. Identify missing values and determine appropriate ways of handling them
4. Identify and investigate duplicated records from the dataset
5. Correct inappropriate and inconsistent data types
6. Identify inconsistent categorical values
7. Investigate potential outliers in the dataset
8. Validate the cleaned dataset
9. Save the cleaned dataset for future analysis

## Data Description

The dataset contains customer-level information from a telecommunication company.
It includes information about:

1. Customer Demographics
2. Services subscribed to
3. Contact Information
4. Payment Methods
5. Customer Tenure
6. Monthly Charges
7. Total Charges
8. Customer Churn Status

**N/B**
The target variable is *Churn* , which indicates whether a customer left the company.
