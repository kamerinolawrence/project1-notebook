#### How many columns have some missing values / blank values

from the above information , we observe that **TotalCharges** columns has the highest number of missing values with **11**.

The rest of the columns have *0* missing values 

*ACTION*

The missing values in the *TotalCharges* column will be further investigated to determine whether they represent genuine missing information or another data quality issue

```python
## Checking affected columns with blank values

df.loc[
    df["TotalCharges"].astype(str).str.strip() == "",
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
]
```

#### Finding

The blank-value check identified **11 blank values in the `TotalCharges` column**.

Further investigation showed that all 11 affected records have a **tenure of 0 months** and valid `MonthlyCharges` values.

Since `TotalCharges` represents the accumulated charges over the customer's tenure, a tenure of 0 indicates that these customers are new accounts with no accumulated charges.

### Cleaning Decision

The 11 blank `TotalCharges` values will be converted to **0** during the data-cleaning stage.

This treatment is based on the observed relationship between `tenure = 0` and the absence of accumulated total charges. The raw dataset will be preserved, and the correction will be applied to the cleaned dataset.
